```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/BIRTH_REG.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum() ##check missing values
```

```python
## cleanning data
df["OBS_VALUE"] = df["OBS_VALUE"].fillna(
    df["OBS_VALUE"].median()
)
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df["Unregistered_Children"] = 100 - df["OBS_VALUE"] ##creating new column
```

```python
df[["INDICATOR_LABEL", "OBS_VALUE", "Unregistered_Children"]].head(10)
```

```python
df["Unregistered_Children"].describe()
```

```python
birth_df = df[df["INDICATOR"] == "BIRTH_REG"].copy()
```

```python
##calculates average birth registration percentage
country_avg = birth_df.groupby("REF_AREA_LABEL")["OBS_VALUE"].mean()

country_avg.head(10)
```

```python
country_avg.sort_values(ascending=False).head(10) ## calculates highest birth reg percentage
```

```python
birth_df["TIME_PERIOD"].unique()
```

```python
yearly_avg = birth_df.groupby("TIME_PERIOD")["OBS_VALUE"].mean()

yearly_avg
```

```python
## Visualization
import matplotlib.pyplot as plt
yearly_avg.sort_index().plot(kind="line", marker="o", figsize=(10,5))

plt.title("Birth Registration Trends Over Time")
plt.xlabel("Year")
plt.ylabel("Average Birth Registration (%)")
plt.grid(True)
plt.show()
```

```python
gender_avg = birth_df[
    birth_df["SEX_LABEL"].isin(["Male", "Female"])
].groupby("SEX_LABEL")["OBS_VALUE"].mean()

gender_avg
```

```python
gender_avg.plot(
    kind="bar",
    title="Birth Registration by Gender",
    xlabel="Gender",
    ylabel="Average Registration Percentage",
    rot=0
)
```

```python
difference = abs(gender_avg["Male"] - gender_avg["Female"])

print("Difference:", round(difference, 2), "%")
```

```python
birth_df["OBS_VALUE"].plot(
    kind="hist",
    bins=20,
    title="Distribution of Birth Registration",
    xlabel="Birth Registration Percentage",
    ylabel="Number of Observations"
)
```

```python
birth_df["OBS_VALUE"].describe()
```

```python
birth_df["Unregistered_Children"] = 100 - birth_df["OBS_VALUE"]

birth_df.to_csv("/content/ID4D_Birth_Registration.csv", index=False)
```
